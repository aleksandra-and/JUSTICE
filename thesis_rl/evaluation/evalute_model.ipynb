## Script to generate evaluation data

Use the cell below by changing `model_dir`, `load_config`, and `evaluation_output_dir` as needed.

Example CLI (same settings used by the notebook cell):
```bash
python ./thesis_rl/evaluation.py \
  --load_config thesis_rl/configs/happo_temp_econ.yaml \
  --model_dir {model_dir} \
  --evaluation_output_dir {evaluation_output_dir}
```

Recommended reward setups:
- HAPPO (12 agents): inverse temperature, net output
- MAPPO (12 agents): inverse temperature, net output

````python
from pathlib import Path
import shlex
import subprocess

def run_evaluation_script(config_path, model_dir, evaluation_output_dir):
    config_path = Path(config_path)
    model_dir = Path(model_dir)
    evaluation_output_dir = Path(evaluation_output_dir)
    evaluation_output_dir.mkdir(parents=True, exist_ok=True)

    cmd = [
        "python",
        "./thesis_rl/evaluation.py",
        "--load_config", str(config_path),
        "--model_dir", str(model_dir),
        "--evaluation_output_dir", str(evaluation_output_dir),
    ]

    # Switch to root directory of the project
    project_root = Path.cwd().resolve().parent.parent
    subprocess.run(cmd, check=True, cwd=project_root, stdout=subprocess.DEVNULL, stderr=subprocess.DEVNULL)
````
