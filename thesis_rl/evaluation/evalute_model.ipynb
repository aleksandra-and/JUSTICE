## Script to generate evaluation data

Use the cell below by changing `model_dir`, `load_config`, and `evaluation_output_dir` as needed.

Example CLI (same settings used by the notebook cell):
```bash
python ./thesis_rl/evaluation.py \
  --load_config thesis_rl/configs/happo_temp_econ.yaml \
  --model_dir {model_dir} \
  --evaluation_output_dir {evaluation_output_dir}
```

Recommended reward setups:
- HAPPO (12 agents): inverse temperature, net output
- MAPPO (12 agents): inverse temperature, net output

````python
from pathlib import Path
import shlex
import subprocess

def run_evaluation_script(config_path, model_dir, evaluation_output_dir):
    config_path = Path(config_path)
    model_dir = Path(model_dir)
    evaluation_output_dir = Path(evaluation_output_dir)
    evaluation_output_dir.mkdir(parents=True, exist_ok=True)

    cmd = [
        "python",
        "./thesis_rl/evaluation.py",
        "--load_config", str(config_path),
        "--model_dir", str(model_dir),
        "--evaluation_output_dir", str(evaluation_output_dir),
    ]

    print("Running:")
    print(" ".join(shlex.quote(part) for part in cmd))

    # Switch to root directory of the project
    project_root = Path.cwd().resolve().parent.parent
    subprocess.run(cmd, check=True, cwd=project_root)
````

### HAPPO evals

````python
# Weight 23, 64, 69, 82, 99
load_config = "thesis_rl/configs/happo_temp_econ.yaml"
model_dir_base_path = "results/happo_mappo_comparison/harl_justice_momarl/harl_justice_momarl/happo"
model_dirs = [ \
    model_dir_base_path + "/momarl_happo_uniform_202603_w23_1235/seed-00001-2026-03-28-03-31-54/models",
    model_dir_base_path + "/momarl_happo_uniform_202603_w64_1235/seed-00001-2026-03-30-06-38-40/models",
    model_dir_base_path + "/momarl_happo_uniform_202603_w69_1235/seed-00001-2026-03-30-07-13-42/models",
    model_dir_base_path + "/momarl_happo_uniform_202603_w82_1235/seed-00001-2026-03-30-02-49-05/models",
    model_dir_base_path + "/momarl_happo_uniform_202603_w99_1235/seed-00001-2026-03-30-09-12-58/models"
]
evaluation_output_dirs = [ \
    "results/detailed_analysis/evaluation/happo_w23",
    "results/detailed_analysis/evaluation/happo_w64",
    "results/detailed_analysis/evaluation/happo_w69",
    "results/detailed_analysis/evaluation/happo_w82",
    "results/detailed_analysis/evaluation/happo_w99"
]

for model_dir, evaluation_output_dir in zip(model_dirs, evaluation_output_dirs):
    print(f"Evaluating model in: \n  {model_dir} \n  and saving results to: \n  {evaluation_output_dir}")
    run_evaluation_script(load_config, model_dir, evaluation_output_dir)
````
